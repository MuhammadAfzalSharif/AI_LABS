```python
# Q1 1)

class GoalBasedAgent:
    def __init__(self, graph, start, goal, depth_limit):
        self.graph = graph
        self.start = start
        self.goal = goal
        self.depth_limit = depth_limit
        self.visited = set()

    def search(self, node, depth):
        if depth > self.depth_limit:
            return None

        if node == self.goal:
            return [node]

        self.visited.add(node)

        for neighbor in self.graph[node]:
            if neighbor not in self.visited:
                path = self.search(neighbor, depth + 1)
                if path:
                    return [node] + path
        return None

    def start_search(self):
        return self.search(self.start, 0)


graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Agent with depth limit of 2
agent = GoalBasedAgent(graph, 'A', 'F', 2)
result = agent.start_search()

print(f"Path found: {result}")
```

```python
#Q1 2)
import heapq

class UtilityBasedAgent:
    def __init__(self, graph, costs, start, goal):
        self.graph = graph
        self.costs = costs
        self.start = start
        self.goal = goal

    def search(self):
        frontier = [(0, self.start, [])]  # (cost, node, path)
        visited = set()

        while frontier:
            cost, node, path = heapq.heappop(frontier)

            if node in visited:
                continue

            visited.add(node)
            new_path = path + [node]

            if node == self.goal:
                return new_path, cost

            for neighbor in self.graph[node]:
                if neighbor not in visited:
                    total_cost = cost + self.costs.get((node, neighbor), 0)
                    heapq.heappush(frontier, (total_cost, neighbor, new_path))

        return None, None

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}
costs = {
    ('A', 'B'): 1, ('A', 'C'): 4,
    ('B', 'A'): 1, ('B', 'D'): 2, ('B', 'E'): 5,
    ('C', 'A'): 4, ('C', 'F'): 3,
    ('D', 'B'): 2,
    ('E', 'B'): 5, ('E', 'F'): 1,
    ('F', 'C'): 3, ('F', 'E'): 1
}

agent = UtilityBasedAgent(graph, costs, 'A', 'F')
path, total_cost = agent.search()

print(f"Path found: {path} with total cost: {total_cost}")
```

```python
# Q2
import itertools

def traveling_salesman_problem(graph):
    num_cities = len(graph)
    min_path_cost = float('inf')
    min_path = None

    for path_perm in itertools.permutations(range(2, num_cities + 1)):
        path = [1] + list(path_perm) + [1]  # Start and end at city 1
        path_cost = 0

        for i in range(len(path) - 1):
            current_city = path[i]
            next_city = path[i+1]
            path_cost += graph[current_city][next_city]

        if path_cost < min_path_cost:
            min_path_cost = path_cost
            min_path = path

    return min_path_cost, min_path

graph = {
    1: {1: 0, 2: 10, 3: 15, 4: 20},
    2: {2: 0, 1: 10, 3: 35, 4: 25},
    3: {3: 0, 1: 15, 2: 35, 4: 30},
    4: {4: 0, 1: 20, 2: 25, 3: 30}
}

min_cost, best_path = traveling_salesman_problem(graph)
print("Shortest Path:", best_path)
print("Minimum Cost:", min_cost)
```

```python
#Q3
graph = {
    1: [4],
    2: [4],
    3: [4],
    4: [1, 2, 3]
}

def dfs(graph, start, goal, depth):
    if depth == 0 and start == goal:
        return [start]
    if depth > 0:
        for neighbor in graph.get(start, []):
            path = dfs(graph, neighbor, goal, depth - 1)
            if path:
                return [start] + path
    return None

def iddfs(graph, start, goal, max_depth):
    for depth in range(max_depth + 1):
        path = dfs(graph, start, goal, depth)
        if path:
            return path
    return None

start_node = 1
goal_node = 3
max_depth = 3
path = iddfs(graph, start_node, goal_node, max_depth)
print(f"Path from {start_node} to {goal_node}: {path}")
```

```python
#Q4
from collections import deque

def bfs(grid, start, goal):
    rows, cols = len(grid), len(grid[0])
    queue = deque([(start, [start])])
    visited = set()

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    while queue:
        (x, y), path = queue.popleft()

        if (x, y) == goal:
            return path

        if (x, y) in visited:
            continue

        visited.add((x, y))

        for dx, dy in directions:
            new_x, new_y = x + dx, y + dy

            if 0 <= new_x < rows and 0 <= new_y < cols and grid[new_x][new_y] != 1:
                queue.append(((new_x, new_y), path + [(new_x, new_y)]))

    return None
grid = [
    [0, 1, 0, 0, 0],
    [0, 1, 0, 1, 0],
    ['S', 0, 0, 1, 'G'],
    [0, 1, 0, 0, 0]
]

start = goal = None
for i in range(len(grid)):
    for j in range(len(grid[0])):
        if grid[i][j] == 'S':
            start = (i, j)
            grid[i][j] = 0
        elif grid[i][j] == 'G':
            goal = (i, j)
            grid[i][j] = 0

# Find the shortest path
shortest_path = bfs(grid, start, goal)
print("Shortest path:", shortest_path)
```
