```python
# Q1 1)

class GoalBasedAgent:
    def __init__(self, graph, start, goal, depth_limit):
        self.graph = graph
        self.start = start
        self.goal = goal
        self.depth_limit = depth_limit
        self.visited = set()

    def search(self, node, depth):
        if depth > self.depth_limit:
            return None

        if node == self.goal:
            return [node]

        self.visited.add(node)

        for neighbor in self.graph[node]:
            if neighbor not in self.visited:
                path = self.search(neighbor, depth + 1)
                if path:
                    return [node] + path
        return None

    def start_search(self):
        return self.search(self.start, 0)


graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Agent with depth limit of 2
agent = GoalBasedAgent(graph, 'A', 'F', 2)
result = agent.start_search()

print(f"Path found: {result}")
```

```python
#Q1 2)
import heapq

class UtilityBasedAgent:
    def __init__(self, graph, costs, start, goal):
        self.graph = graph
        self.costs = costs
        self.start = start
        self.goal = goal

    def search(self):
        frontier = [(0, self.start, [])]  # (cost, node, path)
        visited = set()

        while frontier:
            cost, node, path = heapq.heappop(frontier)

            if node in visited:
                continue

            visited.add(node)
            new_path = path + [node]

            if node == self.goal:
                return new_path, cost

            for neighbor in self.graph[node]:
                if neighbor not in visited:
                    total_cost = cost + self.costs.get((node, neighbor), 0)
                    heapq.heappush(frontier, (total_cost, neighbor, new_path))

        return None, None

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}
costs = {
    ('A', 'B'): 1, ('A', 'C'): 4,
    ('B', 'A'): 1, ('B', 'D'): 2, ('B', 'E'): 5,
    ('C', 'A'): 4, ('C', 'F'): 3,
    ('D', 'B'): 2,
    ('E', 'B'): 5, ('E', 'F'): 1,
    ('F', 'C'): 3, ('F', 'E'): 1
}

agent = UtilityBasedAgent(graph, costs, 'A', 'F')
path, total_cost = agent.search()

print(f"Path found: {path} with total cost: {total_cost}")
```

```python
# Q2
import itertools

distances = {
    'A': {'A': 0, 'B': 10, 'C': 15, 'D': 20},
    'B': {'A': 10, 'B': 0, 'C': 35, 'D': 25},
    'C': {'A': 15, 'B': 35, 'C': 0, 'D': 30},
    'D': {'A': 20, 'B': 25, 'C': 30, 'D': 0}
}

cities = ['A', 'B', 'C', 'D']

def calculate_total_distance(route):
    total_distance = 0
    for i in range(len(route) - 1):
        total_distance += distances[route[i]][route[i+1]]
    total_distance += distances[route[-1]][route[0]]
    return total_distance

def solve_tsp(cities):
    shortest_route = None
    min_cost = float('inf')

    for route in itertools.permutations(cities):
        cost = calculate_total_distance(route)
        if cost < min_cost:
            min_cost = cost
            shortest_route = route

    return shortest_route, min_cost

# Solve the TSP
route, cost = solve_tsp(cities)

print(f"The shortest route is: {route} with a total cost of: {cost}")
```

```python
#Q3
# Graph Representation
graph = {
    1: [4],
    2: [4],
    3: [4],
    4: [1, 2, 3]
}

# Depth-Limited DFS
def dfs(graph, start, goal, depth):
    if depth == 0 and start == goal:
        return [start]
    if depth > 0:
        for neighbor in graph.get(start, []):
            path = dfs(graph, neighbor, goal, depth - 1)
            if path:
                return [start] + path
    return None

# Iterative Deepening DFS
def iddfs(graph, start, goal, max_depth):
    for depth in range(max_depth + 1):
        path = dfs(graph, start, goal, depth)
        if path:
            return path
    return None

# Example usage
start_node = 1
goal_node = 3
max_depth = 3
path = iddfs(graph, start_node, goal_node, max_depth)
print(f"Path from {start_node} to {goal_node}: {path}")
```

```python
#Q4
from collections import deque

def bfs(grid, start, goal):
    rows, cols = len(grid), len(grid[0])
    queue = deque([(start, [start])])
    visited = set()

    directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    while queue:
        (x, y), path = queue.popleft()

        if (x, y) == goal:
            return path

        if (x, y) in visited:
            continue

        visited.add((x, y))

        for dx, dy in directions:
            new_x, new_y = x + dx, y + dy

            if 0 <= new_x < rows and 0 <= new_y < cols and grid[new_x][new_y] != 1:
                queue.append(((new_x, new_y), path + [(new_x, new_y)]))

    return None
grid = [
    [0, 1, 0, 0, 0],
    [0, 1, 0, 1, 0],
    ['S', 0, 0, 1, 'G'],
    [0, 1, 0, 0, 0]
]

start = goal = None
for i in range(len(grid)):
    for j in range(len(grid[0])):
        if grid[i][j] == 'S':
            start = (i, j)
            grid[i][j] = 0
        elif grid[i][j] == 'G':
            goal = (i, j)
            grid[i][j] = 0

# Find the shortest path
shortest_path = bfs(grid, start, goal)
print("Shortest path:", shortest_path)
```
